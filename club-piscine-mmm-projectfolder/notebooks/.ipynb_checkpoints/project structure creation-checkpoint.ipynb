```python
import os
from pathlib import Path

# Define the project root relative to your current notebook location
# Since you're in: /Users/raoul/MMM/Notebook/exploration
# Go up two levels to reach: /Users/raoul/MMM
project_root = Path().cwd().parent.parent  # Adjust if needed
project_name = "club_piscine_mmm"
project_path = project_root / project_name

print(f"Creating project at: {project_path}")

# Create all directories
directories = [
    "src/data",
    "src/features", 
    "src/models",
    "src/analysis",
    "notebooks",
    "config",
    "data/raw",
    "data/interim",
    "data/processed",
    "reports/figures"
]

for dir_path in directories:
    full_path = project_path / dir_path
    full_path.mkdir(parents=True, exist_ok=True)
    print(f"✓ Created: {full_path}")

# Create empty files
files = [
    "requirements.txt",
    "README.md",
    ".gitignore",
    "src/__init__.py",
    "src/data/__init__.py",
    "src/features/__init__.py",
    "src/models/__init__.py",
    "src/analysis/__init__.py"
]

for file_name in files:
    file_path = project_path / file_name
    file_path.touch()
    print(f"✓ Created: {file_path}")

print(f"\n✅ Project structure created successfully!")
print(f"📁 Location: {project_path}")
print(f"\nNow add your client files to: {project_path}/data/raw/")
```
